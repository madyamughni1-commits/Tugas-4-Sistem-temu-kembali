## 1. Import Library

```python
import re
from Sastrawi.StopWordRemover.StopWordRemoverFactory import StopWordRemoverFactory
from Sastrawi.Stemmer.StemmerFactory import StemmerFactory

print("Library berhasil diimport.")
```

## 2. Inisialisasi Stopword & Stemmer Sastrawi

```python
stopword_factory = StopWordRemoverFactory()
stopwords = set(stopword_factory.get_stop_words())
stemmer = StemmerFactory().create_stemmer()

print(f"Jumlah stopwords Sastrawi: {len(stopwords)}")
```

## 3. Fungsi `preprocess_text`

Pipeline:
1. Case folding
2. Cleaning (hapus angka, tanda baca, karakter khusus)
3. Tokenisasi
4. Stopwords removal (Sastrawi)
5. Stemming (Sastrawi)

```python
def preprocess_text(text):
    text = text.lower()
    text = re.sub(r'[^a-z\s]', ' ', text)
    text = re.sub(r'\s+', ' ', text).strip()

    tokens = text.split()

    tokens = [token for token in tokens if token not in stopwords]

    tokens = [stemmer.stem(token) for token in tokens]

    tokens = [token for token in tokens if token.strip() != '']

    return tokens
```

## 4. Dataset (5 Dokumen Bahasa Indonesia)

Tema: teknologi, komputer, dan pendidikan.

```python
documents = [
    "Pada tahun 2024, Kecerdasan Buatan (AI) semakin berkembang pesat di Indonesia. "
    "Banyak universitas mengembangkan penelitian tentang machine learning dan deep learning.",

    "Jaringan komputer terdiri dari beberapa perangkat yang saling terhubung. "
    "Topologi jaringan dapat berupa star, bus, atau ring.",

    "Sistem operasi mengelola sumber daya komputer seperti memori, prosesor, dan perangkat input-output. "
    "Linux dan Windows adalah contoh sistem operasi.",

    "Basis data menyimpan data dalam tabel yang saling berhubungan. "
    "Query SQL digunakan untuk mengambil dan memanipulasi data.",

    "Sistem temu kembali informasi mencari dokumen relevan berdasarkan query pengguna. "
    "Evaluasi dilakukan dengan precision dan recall."
]

print(f"Total dokumen: {len(documents)}")
```

## 5. Terapkan Preprocessing pada Ke-5 Dokumen
Menampilkan **sebelum** vs **sesudah** preprocessing secara lengkap.

```python
hasil = []  # menyimpan hasil untuk analisis

for i, doc in enumerate(documents, 1):
    tokens_sebelum = doc.split()
    tokens_sesudah = preprocess_text(doc)
    hasil.append((tokens_sebelum, tokens_sesudah))

    print("=" * 100)
    print(f"DOKUMEN {i}")
    print("=" * 100)
    print("Teks Asli   :", doc)
    print()
    print("Sebelum     :", tokens_sebelum)
    print()
    print("Sesudah     :", tokens_sesudah)
    print()
```

## 6. Statistik: Jumlah Token & Persentase Pengurangan

```python
print("=" * 100)
print("PERBANDINGAN JUMLAH TOKEN")
print("=" * 100)
print(f"{'Dokumen':<10}{'Sebelum':<12}{'Sesudah':<12}{'Pengurangan (%)':<20}")
print("-" * 100)

total_before = 0
total_after = 0

for i, (before, after) in enumerate(hasil, 1):
    pengurangan = (len(before) - len(after)) / len(before) * 100
    total_before += len(before)
    total_after += len(after)
    print(f"D{i:<9}{len(before):<12}{len(after):<12}{pengurangan:<20.2f}")

print("-" * 100)
pengurangan_total = (total_before - total_after) / total_before * 100
print(f"{'TOTAL':<10}{total_before:<12}{total_after:<12}{pengurangan_total:<20.2f}")
```

## 7. Analisis Singkat

**Analisis Dampak Preprocessing terhadap Kualitas Data untuk Sistem IR:**

Preprocessing berhasil mereduksi jumlah token secara signifikan karena menghilangkan 
angka (seperti "2024"), tanda baca, karakter khusus (seperti tanda kurung dan tanda hubung), 
serta stopwords umum Bahasa Indonesia (seperti "yang", "dan", "di", "dari"). 
Proses stemming dengan Sastrawi juga mengubah kata berimbuhan menjadi bentuk dasarnya 
(misalnya "mengembangkan" → "kembang", "menyimpan" → "simpan"), sehingga variasi bentuk 
kata dari akar yang sama dapat disatukan. Dampaknya bagi sistem temu kembali informasi (IR): 
(1) representasi dokumen menjadi lebih ringkas dan terfokus pada kata bermakna; 
(2) mengurangi noise sehingga perhitungan kemiripan (cosine similarity) lebih akurat; 
(3) meningkatkan recall karena query dan dokumen lebih mudah dicocokkan pada level kata dasar. 
Namun, stemming agresif dapat menghilangkan nuansa makna tertentu, sehingga perlu 
dipertimbangkan trade-off antara kesederhanaan representasi dan kekayaan semantik.
